```python
# RFM Segmentation (Recency, Frequency, Monetary)
query_rfm = """
WITH customer_rfm AS (
    SELECT
        LYLTY_CARD_NBR,
        MAX(DATE) AS last_purchase,
        COUNT(DISTINCT TXN_ID) AS frequency,
        SUM(TOT_SALES) AS monetary
    FROM df
    GROUP BY LYLTY_CARD_NBR
),
recency_calc AS (
    SELECT
        LYLTY_CARD_NBR,
        date_diff('day', last_purchase, (SELECT MAX(DATE) FROM df)) AS recency,
        frequency,
        monetary
    FROM customer_rfm
)
SELECT
    LYLTY_CARD_NBR,
    recency,
    frequency,
    monetary
FROM recency_calc
ORDER BY monetary DESC
LIMIT 20
"""
df_rfm = duckdb.query(query_rfm).to_df()
df_rfm
```

**Calculating RFM Scores**

```python
# Calculating RFM Scores
query_rfmscore = """
WITH rfm AS (
    SELECT
        LYLTY_CARD_NBR,
        MAX(DATE) AS last_purchase,
        COUNT(DISTINCT TXN_ID) AS Frequency,
        SUM(TOT_SALES) AS Monetary,
        date_diff('day', MAX(DATE), (SELECT MAX(DATE) FROM df)) AS Recency
    FROM df
    GROUP BY LYLTY_CARD_NBR
),
rfm_ranked AS (
    SELECT
        LYLTY_CARD_NBR,
        Recency,
        Frequency,
        Monetary,

        (6 - NTILE(5) OVER (ORDER BY Recency ASC)) AS R,
        NTILE(5) OVER (ORDER BY Frequency ASC) AS F,
        NTILE(5) OVER (ORDER BY Monetary ASC) AS M
    FROM rfm
)
SELECT
    LYLTY_CARD_NBR,
    Recency,
    Monetary,
    Frequency,
    R, F, M,
    CAST(R AS VARCHAR) || CAST(F AS VARCHAR) || CAST(M AS VARCHAR) AS RFM_Score
FROM rfm_ranked
ORDER BY RFM_Score DESC
"""

df_rfmscore = duckdb.query(query_rfmscore).to_df()
df_rfmscore
```

**Customer Segmentation From Recency and Frequency**

```python
# Customer Segmentation From Recency and Frequency
query_rfmseg = """
WITH rfm AS (
    SELECT
        LYLTY_CARD_NBR,
        MAX(DATE) AS last_purchase,
        COUNT(DISTINCT TXN_ID) AS Frequency,
        SUM(TOT_SALES) AS Monetary,
        date_diff('day', MAX(DATE), (SELECT MAX(DATE) FROM df)) AS Recency
    FROM df
    GROUP BY LYLTY_CARD_NBR
),
rfm_ranked AS (
    SELECT
        LYLTY_CARD_NBR,
        Recency,
        Frequency,
        Monetary,
        (6 -NTILE(5) OVER (ORDER BY Recency ASC)) AS R,
        NTILE(5) OVER (ORDER BY Frequency ASC) AS F,
        NTILE(5) OVER (ORDER BY Monetary ASC) AS M
    FROM rfm
),
rfm_score AS (
    SELECT
        LYLTY_CARD_NBR,
        R, F, M,
        CAST(R AS VARCHAR) || CAST(F AS VARCHAR) AS RF,
        CAST(R AS VARCHAR) || CAST(F AS VARCHAR) || CAST(M AS VARCHAR) AS RFM_Score
        FROM rfm_ranked
)
SELECT
    LYLTY_CARD_NBR,
    R,F,M,
    RFM_Score,
    CASE
        WHEN RF ~ '^[1-2][1-2]$' THEN 'Hibernating'
        WHEN RF ~ '^[1-2][3-4]$' THEN 'At Risk'
        WHEN RF ~ '^[1-2]5$'     THEN 'Can''t lose'
        WHEN RF ~ '^3[1-2]$'      THEN 'About to Sleep'
        WHEN RF = '33'           THEN 'Need Attention'
        WHEN RF ~ '^[3-4][4-5]$'  THEN 'Loyal Customers'
        WHEN RF = '41'           THEN 'Promising'
        WHEN RF = '51'           THEN  'New Customers'
        WHEN RF ~ '^[4-5][2-3]$' THEN 'Potential Loyalists'
        WHEN RF ~ '^5[4-5]$'     THEN 'Champions'
        ELSE 'Others'
    END AS Segment
FROM rfm_score
ORDER BY Segment;
"""
df_rfmseg = duckdb.query(query_rfmseg).to_df()
df_rfmseg
```

**Calculating the Average Value of Each Segment**

```python
# Calculating the Average Value of Each Segment
query_segavg = """
WITH rfm AS (
    SELECT
        LYLTY_CARD_NBR,
        MAX(DATE) AS last_purchase,
        COUNT(DISTINCT TXN_ID) AS Frequency,
        SUM(TOT_SALES) AS Monetary,
        date_diff('day', MAX(DATE), (SELECT MAX(DATE) FROM df)) AS Recency
    FROM df
    GROUP BY LYLTY_CARD_NBR
),

rfm_ranked AS (
    SELECT
        LYLTY_CARD_NBR,
        Recency,
        Frequency,
        Monetary,
        (6 - NTILE(5) OVER (ORDER BY Recency ASC)) AS R,
        NTILE(5) OVER (ORDER BY Frequency ASC) AS F,
        NTILE(5) OVER (ORDER BY Monetary ASC) AS M
    FROM rfm
),

rfm_score AS (
    SELECT
        LYLTY_CARD_NBR,
        Recency,
        Frequency,
        Monetary,
        CAST(R AS VARCHAR) || CAST(F AS VARCHAR) AS RF,
        CASE
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^[1-2][1-2]$' THEN 'Hibernating'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^[1-2][3-4]$' THEN 'At Risk'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^[1-2]5$'     THEN 'Can''t Lose'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^3[1-2]$'     THEN 'About to Sleep'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '33'           THEN 'Need Attention'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^[3-4][4-5]$' THEN 'Loyal Customers'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '41'           THEN 'Promising'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '51'           THEN 'New Customers'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^[4-5][2-3]$' THEN 'Potential Loyalists'
            WHEN (CAST(R AS VARCHAR) || CAST(F AS VARCHAR)) ~ '^5[4-5]$'     THEN 'Champions'
            ELSE 'Others'
        END AS Segment
    FROM rfm_ranked
),

segment_metrics AS (
    SELECT
        Segment,
        COUNT(DISTINCT LYLTY_CARD_NBR) AS num_customers,
        AVG(Recency)   AS avg_recency,
        AVG(Frequency) AS avg_frequency,
        AVG(Monetary)  AS avg_monetary
    FROM rfm_score
    GROUP BY Segment
)

SELECT
    Segment,
    num_customers,

    NTILE(5) OVER (ORDER BY avg_recency DESC)  AS recency_score,
    NTILE(5) OVER (ORDER BY avg_frequency ASC) AS frequency_score,
    NTILE(5) OVER (ORDER BY avg_monetary ASC)  AS monetary_score,

    ROUND(avg_recency, 2)   AS avg_recency,
    ROUND(avg_frequency, 2) AS avg_frequency,
    ROUND(avg_monetary, 2)  AS avg_monetary,

    ROUND(
        (num_customers * 100.0) / SUM(num_customers) OVER(),
        1
    ) AS pct_customers
FROM segment_metrics
ORDER BY monetary_score DESC;

"""

df_segavg2 = duckdb.query(query_segavg).to_df()
df_segavg2
```

**Help function (find correlation and magnitude)**

```python
# 4) Correlation function
def calculate_correlation(df, metricCol, storeComparison):
    rows = []
    # months for this trial store (should be complete)
    trial_series = df[df['STORE_NBR'] == storeComparison][['YEARMONTH', metricCol]]
    for s in df['STORE_NBR'].unique():
        if s == storeComparison:
            continue
        control_series = df[df['STORE_NBR'] == s][['YEARMONTH', metricCol]]
        merged = pd.merge(trial_series, control_series, on='YEARMONTH', how='inner', suffixes=('_trial','_control'))
        # require all months present
        if len(merged) == 0:
            continue
        corr = np.nan
        try:
            corr, _ = pearsonr(merged[f"{metricCol}_trial"], merged[f"{metricCol}_control"])
        except Exception:
            corr = np.nan
        rows.append({'Store1': storeComparison, 'Store2': s, 'corr_measure': corr})
    return pd.DataFrame(rows)
```

```python
# 5) Magnitude distance function
def calculate_magnitude_distance(df, metricCol, storeComparison):
    # build calcDistTable rows of (Store1, Store2, YEARMONTH, measure)
    calc_rows = []
    storeNumbers = df['STORE_NBR'].unique()
    # trial series by month
    #trial_by_month = df[df'STORE_NBR'] == storeComparison][['YEARMONTH', metricCol]].set_index('YEARMONTH')
    trial_by_month = df[df['STORE_NBR'] == storeComparison][['YEARMONTH', metricCol]].set_index('YEARMONTH')
    for s in storeNumbers:
        # get control series
        control_by_month = df[df['STORE_NBR'] == s][['YEARMONTH', metricCol]].set_index('YEARMONTH')
        # align months present in trial_by_month
        months = trial_by_month.index.intersection(control_by_month.index)
        for m in months:
            trial_val = trial_by_month.at[m, metricCol]
            control_val = control_by_month.at[m, metricCol]
            measure = abs(trial_val - control_val)
            calc_rows.append({'Store1': storeComparison, 'Store2': s, 'YEARMONTH': m, 'measure': measure})
    calcDistTable = pd.DataFrame(calc_rows)
    if calcDistTable.empty:
        return pd.DataFrame(columns=['Store1','Store2','mag_measure'])
    # compute min/max per Store1 & YEARMONTH (same grouping as R)
    minMax = calcDistTable.groupby(['Store1','YEARMONTH'])['measure'].agg(minDist='min', maxDist='max').reset_index()
    # merge
    distTable = calcDistTable.merge(minMax, on=['Store1','YEARMONTH'], how='left')
    # avoid division by zero
    def magnitudeMeasureRow(r):
        denom = (r['maxDist'] - r['minDist'])
        if pd.isna(denom) or denom == 0:
            return 1.0  # if no variation that month, treat as max similarity
        return 1 - (r['measure'] - r['minDist']) / denom
    distTable['magnitudeMeasure'] = distTable.apply(magnitudeMeasureRow, axis=1)
    # finalDistTable: average magnitudeMeasure across months per Store1,Store2
    final = distTable.groupby(['Store1','Store2'])['magnitudeMeasure'].mean().reset_index().rename(columns={'magnitudeMeasure':'mag_measure'})
    return final
```

```python
# 6) Function to compute final control score merge/weights
def compute_control_scores(trial_store):
    # using totSales and nCustomers (R template used these two to choose controls)
    # compute correlation & magnitude for both metrics
    corr_sales = calculate_correlation(preTrialMeasures, 'totSales', trial_store)
    corr_cust  = calculate_correlation(preTrialMeasures, 'nCustomers', trial_store)
    mag_sales  = calculate_magnitude_distance(preTrialMeasures, 'totSales', trial_store)
    mag_cust   = calculate_magnitude_distance(preTrialMeasures, 'nCustomers', trial_store)

    # merge corr and mag for each driver
    score_sales = pd.merge(corr_sales, mag_sales, on=['Store1','Store2'], how='inner')
    score_sales['scoreNSales'] = 0.5 * score_sales['corr_measure'] + 0.5 * score_sales['mag_measure']

    score_cust = pd.merge(corr_cust, mag_cust, on=['Store1','Store2'], how='inner')
    score_cust['scoreNCust'] = 0.5 * score_cust['corr_measure'] + 0.5 * score_cust['mag_measure']

    # combine drivers into final score
    final = pd.merge(score_sales[['Store2','scoreNSales']], score_cust[['Store2','scoreNCust']], on='Store2', how='inner')
    final['finalControlScore'] = 0.5 * (final['scoreNSales'] + final['scoreNCust'])
    # remove self if present
    final = final[final['Store2'] != trial_store].copy()
    final = final.sort_values('finalControlScore', ascending=False).reset_index(drop=True)
    return final
```

```python
# 7) Pre-trial & trial check function that returns plots & metrics
def evaluate_trial_vs_control(trial_store, control_store, show_plots=True):
    # pre-trial data (months < PRE_TRIAL_CUTOFF)
    pre = measureOverTime[(measureOverTime['YEARMONTH'] < PRE_TRIAL_CUTOFF) & (measureOverTime['STORE_NBR'].isin([trial_store, control_store]))].copy()
    # trial data (TRIAL_MONTHS)
    trial = measureOverTime[measureOverTime['YEARMONTH'].isin(TRIAL_MONTHS) & (measureOverTime['STORE_NBR'].isin([trial_store, control_store]))].copy()

    # pre-trial correlation & percent diff series
    pre_trial_trial = pre[pre['STORE_NBR'] == trial_store][['YEARMONTH','totSales']].set_index('YEARMONTH')
    pre_trial_ctrl  = pre[pre['STORE_NBR'] == control_store][['YEARMONTH','totSales']].set_index('YEARMONTH')
    common_months = pre_trial_trial.index.intersection(pre_trial_ctrl.index)
    merged_pre = pd.DataFrame({
        'YEARMONTH': list(common_months),
        'trial': pre_trial_trial.loc[common_months,'totSales'].values,
        'control': pre_trial_ctrl.loc[common_months,'totSales'].values
    })
    # Pearson r for pre-trial
    r_pre, _ = pearsonr(merged_pre['trial'], merged_pre['control'])
    # percentageDiff = abs(trial - scaled_control)/scaled_control (we will scale later)
    # scale control to trial on pre-trial totals
    sum_trial_pre = pre_trial_trial.loc[common_months,'totSales'].sum()
    sum_ctrl_pre  = pre_trial_ctrl.loc[common_months,'totSales'].sum()
    scaling_factor = sum_trial_pre / (sum_ctrl_pre + 1e-9)
    # create scaled control for trial months
    ctrl_trial = trial[trial['STORE_NBR'] == control_store][['YEARMONTH','totSales']].set_index('YEARMONTH')
    ctrl_trial = ctrl_trial.reindex(TRIAL_MONTHS).fillna(0)
    ctrl_trial['scaled'] = ctrl_trial['totSales'] * scaling_factor
    trial_trial = trial[trial['STORE_NBR'] == trial_store][['YEARMONTH','totSales']].set_index('YEARMONTH').reindex(TRIAL_MONTHS).fillna(0)

    merged_trial = pd.DataFrame({
        'YEARMONTH': TRIAL_MONTHS,
        'trial_sales': trial_trial['totSales'].values,
        'control_scaled': ctrl_trial['scaled'].values
    })
    # compute percent change and simple t-test (Welch)
    pct_change = (merged_trial['trial_sales'].mean() - merged_trial['control_scaled'].mean()) / (merged_trial['control_scaled'].mean() + 1e-9) * 100
    try:
        tstat, pval = ttest_ind(merged_trial['trial_sales'], merged_trial['control_scaled'], equal_var=False)
    except Exception:
        tstat, pval = np.nan, np.nan

    # pre-trial percent diffs for stdDev as per R:
    # build percentageDiff = abs(control_scaled_pre - trial_pre)/control_scaled_pre for pre-trial months
    ctrl_pre = pre[pre['STORE_NBR'] == control_store][['YEARMONTH','totSales']].set_index('YEARMONTH')
    # align months and compute scaled control values
    aligned_months = common_months
    ctrl_pre_aligned = ctrl_pre.loc[aligned_months,'totSales'] * scaling_factor
    trial_pre_aligned = pre_trial_trial.loc[aligned_months,'totSales']
    percentageDiff_pre = (abs(ctrl_pre_aligned - trial_pre_aligned) / (ctrl_pre_aligned + 1e-9)).values
    stdDev = np.nanstd(percentageDiff_pre, ddof=1) if len(percentageDiff_pre) > 1 else np.nan

    # compute t-values for trial months using R test formula (tValue = (x - u)/stdDev)
    # where x is percentageDiff for trial months and u=0 (null = 0 difference)
    pctdiff_trial_months = (abs(merged_trial['control_scaled'] - merged_trial['trial_sales']) / (merged_trial['control_scaled'] + 1e-9)).values
    t_values = (pctdiff_trial_months - 0) / (stdDev + 1e-9) if not np.isnan(stdDev) else np.full_like(pctdiff_trial_months, np.nan)

    # plots
    if show_plots:
        # pre-trial trend
        plt.figure(figsize=(12,4))
        sns.lineplot(data=pre, x='YEARMONTH', y='totSales', hue='STORE_NBR')
        plt.title(f'Pre-trial totSales: trial {trial_store} vs control {control_store}')
        plt.xticks(rotation=45)
        plt.show()

        # pre-trial scatter
        plt.figure(figsize=(5,4))
        plt.scatter(merged_pre['trial'], merged_pre['control'])
        plt.xlabel('trial totSales (pre)')
        plt.ylabel('control totSales (pre)')
        plt.title(f'Pre-trial correlation r={r_pre:.3f}')
        plt.show()

        # trial comparison
        plt.figure(figsize=(10,4))
        plt.plot(merged_trial['YEARMONTH'], merged_trial['trial_sales'], marker='o', label=f'trial {trial_store}')
        plt.plot(merged_trial['YEARMONTH'], merged_trial['control_scaled'], marker='o', label=f'control {control_store} (scaled)')
        plt.fill_between(merged_trial['YEARMONTH'], merged_trial['control_scaled']*(1-2*stdDev if not np.isnan(stdDev) else 0.95),
                         merged_trial['control_scaled']*(1+2*stdDev if not np.isnan(stdDev) else 1.05),
                         color='gray', alpha=0.2, label='±2*stdDev (pre-trial pct diffs)')
        plt.legend(); plt.title('Trial period: scaled control vs trial'); plt.show()

    return {
        'trial_store': trial_store,
        'control_store': control_store,
        'pre_corr': float(r_pre),
        'scaling_factor': float(scaling_factor),
        'pct_change': float(pct_change),
        'tstat': float(tstat) if not np.isnan(tstat) else np.nan,
        'pval': float(pval) if not np.isnan(pval) else np.nan,
        'stdDev_pre_pctdiff': float(stdDev)
    }
```
